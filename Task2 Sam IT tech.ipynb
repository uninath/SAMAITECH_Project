TASK 2: CUISINE COMBINATION
Dataset: Zomato Restaurants Dataset(Kaggle)
Goal: Identifying the most common combinations of cuisines in the dataset and to determine if certain cuisines combinations tend to have higher ratings.

Step 1: Load the dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Admin\Desktop\zomato.csv", encoding='latin-1')
df.head()
```

Step 2: Most commom cuisine combination

```python
Cuisine_combo_counts = df['Cuisines'].value_counts().head(15)
Cuisine_combo_counts
```

```python
print(df.columns.tolist())
```

Step 3: Average rating per cuisine combination

```python
Cuisine_ratings = df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False)
Cuisine_ratings.head(15)
```

Problem: First attempt without filtering reveals a common data pitfall. The top results all show a perfect 4.9 rating which is misleading.
To fix: filter combinations with a reasonable sample size(at least 20 resturants).

```python
combo_stats = df.groupby('Cuisines').agg(
    avg_rating=('Aggregate rating', 'mean'),
    restaurant_count=('Aggregate rating', 'count')
)

# only keep combos with at least 20 restaurants
reliable_combos = combo_stats[combo_stats['restaurant_count'] >=20].sort_values('avg_rating', ascending=False)
reliable_combos.head(15)
```

CONCLUSION
The most popular cuisine combinations are not necessarily the highest-rated ones. Single-cuisine, internationally-themed restaurants tends to outperform broad multi-cuisine combinations in customer satisfaction.
